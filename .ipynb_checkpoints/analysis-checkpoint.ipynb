```python
from scripts.gtf_utils import parse_gene_map

import numpy as np
import scanpy as sc
import matplotlib.pyplot as plt
import re

adata_dir = "anndata/pbmc_1k_v3.h5ad"
adata = sc.read_h5ad(adata_dir)
print(f"Loaded h5ad file from `{adata_dir}`")
```

```python
print(adata)
```

```python
# Replacing with symbol names as its a bit easier to work with
gene_map = parse_gene_map("data/ref/genes.gtf")

bare_map = {re.sub(r"\.\d+$", "", k): v for k, v in gene_map.items()}
bare_ids = adata.var_names.str.replace(r"\.\d+$", "", regex=True)

adata.var["ensembl_id"] = adata.var_names
adata.var["gene_symbol"] = [bare_map.get(g, (g, ""))[0] for g in bare_ids]
adata.var_names = adata.var["gene_symbol"].values
adata.var_names_make_unique()

# Mito flagging, mirroring from scripts/plots.py
adata.var["chromosome"] = [bare_map.get(g, (g, ""))[1] for g in bare_ids]
adata.var["mt"] = adata.var["chromosome"] == "chrM"
print(f"{adata.var['mt'].sum()} flagged as mito genes")

sc.pp.calculate_qc_metrics(adata, qc_vars=["mt"], inplace=True, percent_top=None)
```

```python
print(f"Before filtering {adata.shape}")

# Filtering
sc.pp.filter_cells(adata, min_genes=200)  # removes droplets with little ambient RNA
sc.pp.filter_genes(adata, min_cells=3)    # removes uninformative genes 

adata = adata[adata.obs["total_counts"] < 50000, :] # removes outliers (plots/pbmc_1k_V3/counts_vs_genes.png)
adata = adata[adata.obs["pct_counts_mt"] < 20, :]

print(f"After filtering {adata.shape}")
```

```python
# Log normalization
adata.layers["counts"] = adata.X.copy()
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)
```

```python
# Graphing PCA variance ratio for UMAP
sc.tl.pca(adata, random_state=0)
sc.pl.pca_variance_ratio(adata, n_pcs=50, log=True)
```

```python
# Computing and plotting UMAP
sc.pp.neighbors(adata, n_pcs=15, random_state=0)
sc.tl.umap(adata, random_state=0)
sc.tl.leiden(adata, resolution=1.0, key_added="leiden_1.0", random_state=0)

sc.pl.umap(adata, color="leiden_1.0")
```

```python
# Okay, lets take a look at these clusters
# Markers per scanpy tutorial convention
pbmc_markers = {
    "T cells":       ["CD3D", "CD3E"],
    "CD4 T":         ["CD4", "IL7R"],
    "CD8 T":         ["CD8A", "CD8B"],
    "B cells":       ["MS4A1", "CD79A"],
    "NK cells":      ["NKG7", "GNLY"],
    "Monocytes":     ["CD14", "LYZ"],
    "FCGR3A+ Mono":  ["FCGR3A"],
    "Dendritic":     ["FCER1A"],
    "Platelets":     ["PPBP"],
}

genes = [g for gs in pbmc_markers.values() for g in gs if g in adata.var_names]

sc.pl.umap(adata, color=genes, ncols=3, cmap="viridis")
```

```python
for label, genes in pbmc_markers.items():
    present = [g for g in genes if g in adata.var_names]
    if present:
        sc.tl.score_genes(adata, present, score_name=f"score_{label}", random_state=0)

score_cols = [c for c in adata.obs.columns if c.startswith("score_")]

# average score per cluster per signature
cluster_scores = adata.obs.groupby("leiden_1.0")[score_cols].mean()
cluster_scores.columns = [c.replace("score_", "") for c in cluster_scores.columns]
cluster_scores["top_signature"] = cluster_scores.idxmax(axis=1)

adata.obs["cell_type"] = adata.obs["leiden_1.0"].map(cluster_scores["top_signature"])
sc.pl.umap(adata, color="cell_type", legend_loc="on data")
```
